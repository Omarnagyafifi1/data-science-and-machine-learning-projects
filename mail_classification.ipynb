```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report

X=df['message']
y=df['label']
X_train,X_test,ytrain,ytest=train_test_split(X,y,test_size=0.2,random_state=4)

lb=LabelEncoder()
ytrain= lb.fit_transform(ytrain)
ytest=lb.transform(ytest)

# Instantiate TfidfVectorizer as 'vectorizer' to avoid overwriting the class name
vectorizer=TfidfVectorizer(min_df=1,stop_words='english')
X_train=vectorizer.fit_transform(X_train).toarray()
X_test=vectorizer.transform(X_test).toarray()

model=MultinomialNB()
model.fit(X_train,ytrain)
ypred=model.predict(X_test)
print(classification_report(ytest,ypred))

import joblib
# Save the model
joblib.dump(model,'mailclassifier.joblib')
# Save the TfidfVectorizer as well, it's crucial for future predictions
joblib.dump(vectorizer, 'tfidf_vectorizer.joblib')
# Also save the LabelEncoder to decode predictions back to original labels
joblib.dump(lb, 'label_encoder.joblib')
```

```python
import gradio as gr
import joblib

# Load the trained model, TF-IDF vectorizer, and LabelEncoder
model = joblib.load('mailclassifier.joblib')
vectorizer = joblib.load('tfidf_vectorizer.joblib')
label_encoder = joblib.load('label_encoder.joblib')

def predict_spam(text):
    # Transform the input text using the loaded vectorizer
    text_vectorized = vectorizer.transform([text]).toarray()

    # Make prediction using the loaded model
    prediction = model.predict(text_vectorized)

    # Decode the prediction back to original labels (ham/spam)
    predicted_label = label_encoder.inverse_transform(prediction)[0]

    return predicted_label

# Create the Gradio interface
iface = gr.Interface(fn=predict_spam,
                     inputs=gr.Textbox(lines=5, placeholder="Enter your message here..."),
                     outputs="text",
                     title="SMS Spam Classifier",
                     description="Enter an SMS message to classify it as 'ham' (not spam) or 'spam'.")

# Launch the interface
iface.launch(debug=True)
```
